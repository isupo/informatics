## 1. Что такое функция?

Функция — это кусочек программы, у которого есть **имя**. Мы пишем его один раз, а потом можем «дёргать» за это имя сколько угодно раз.

Вот самая простая функция:

```python
def say_hello():
    print("Привет!")
```

Что здесь к чему:
- `def` — ключевое слово, оно говорит Python: «сейчас я объявляю функцию»;
- `say_hello` — это имя функции (мы его придумали);
- `()` — здесь пишутся параметры (пока их нет);
- `:` — двоеточие в конце строки;
- всё, что с отступом — это **тело** функции. Оно выполняется, когда функцию вызовут.

Чтобы функция сработала, её нужно **вызвать** — написать её имя со скобками:

```python
say_hello()
```

````python
def say_hello():
    print("Привет!")

say_hello()
say_hello()
say_hello()
````

## 2. Функция, которая возвращает результат

Печатать — это хорошо, но чаще всего функция должна **вернуть** значение, чтобы мы могли его использовать дальше. Для этого есть слово `return`.

```python
def double(x):
    return x * 2

result = double(5)
print(result)   # 10
```

Как только Python доходит до `return`, функция сразу завершается и отдаёт значение тому, кто её вызвал.

````python
def double(x):
    return x * 2

print(double(5))
print(double(100))
print(double(0))
````

## 3. Параметры и аргументы

**Параметры** — это имена в скобках при объявлении функции. **Аргументы** — это конкретные значения, которые мы передаём при вызове.

```python
def add(a, b):        # a и b — параметры
    return a + b

print(add(2, 3))      # 2 и 3 — аргументы
```

Параметров может быть сколько угодно. Имя функции и имена параметров придумываешь ты сам(а), но они должны быть понятными.

````python
def introduce(name, age):
    return f"Меня зовут {name}, мне {age} лет"

print(introduce("Аня", 12))
print(introduce("Петя", 13))
````

## 4. Значения по умолчанию

Иногда удобно, чтобы у параметра уже было готовое значение. Тогда при вызове его можно **не передавать** — подставится значение по умолчанию.

```python
def greet(name="друг"):
    return f"Привет, {name}!"

print(greet())           # Привет, друг!
print(greet("Аня"))      # Привет, Аня!
```

Важно: параметры **со значением по умолчанию** должны идти **после** обычных параметров.

````python
def power(base, exponent=2):
    return base ** exponent

print(power(3))       # 3 в квадрате = 9
print(power(2, 3))    # 2 в кубе = 8
print(power(5, 1))    # 5
````
